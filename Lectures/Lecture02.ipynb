Use the `sympy` package to compute the exact value for math problem.

````python
# define some variables, x, y, z, t
from sympy import symbols, Function
````

````python
# define x var
x = symbols("x")
````

````python
# we can from a quad function
f = 2 * x**2 -3*x + 1
````

````python
f
````

````python
from sympy import factor
````

````python
# factor the quad function f
factor(f)
````

````python
# we can define complex functions
from sympy import sin, exp, cos, tan, ln
````

````python
# define g = e^(2x)*sin(3x) + ln(x)
g = exp(2*x)*sin(3*x)+ln(x)
````

````python
g
````

## Derivates of function

````python
from sympy import diff
````

````python
diff(g,x)
````

````python
# 2nd derivative
diff(g, x, 2)
````

````python
# 10th derivatice
diff(g, x, 10)
````

## Integral

````python
from sympy import integrate
````

**Example:** Compute the integral
- $\int_{0}^1(\sin(x)e^{2x}dx)$
- $\int_{-\infty}^{\infty}e^{-x^2}dx$

Syntax for integral
```python
integrate(f, (var, start, end))
```

````python
# compute the first integral
integrate(x*sin(x)*exp(2*x), (x, 0, 1))
````
