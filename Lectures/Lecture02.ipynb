### Taylor series

Given a function $y=f(x)$ and $x=a$, we can compute the Taylor series
$$
T_n(x) = \sum_{k=0}^n\frac{f^{(k)}(a)}{k!}(x-a)^k
$$

````python
# import series from sympy
from sympy import series
````

**Example:** Compute the Taylor series of $y = e^{x^2}\sin(x)$ and $x=0$ up to order $n=10$.

````python
from sympy import symbols, exp, sin, series
````

````python
x = symbols("x")
````

````python
f = exp(x**2)*sin(x)
````

````python
f
````

````python
a = 0
n = 10
````

````python
f_series = series(f, x, a, n)
````

````python
f_series
````

````python
# we can remove the O()
t_series = f_series.removeO()
````

````python
t_series
````

## Partial derivates and Maxima, Minima

**Example:** Given the function
$$
f(x,y) = x^3 + y^3 -3 xy
$$

a) Find partial derivatives of the function

b) Find the critical points of the function

c) Compute the hessian of the function

d) Classify critical points (local max, local min, saddle points)

````python
from sympy import symbols, diff, hessian
````

````python
# define x and y varialbes
# x = symbols("x")
# y = symbols("y")
x, y = symbols("x y")
````

````python
f = x**3 + y**3 - 3*x*y
f
````

Compute the derivative wrt $x$:
$$
\frac{\partial f}{\partial x}
$$

````python
f_x = diff(f, x)
f_x
````

````python
# or
f_x = f.diff(x)
f_x
````

````python
# compute the derivative wrt y
f_y = diff(f, y)
f_y
````

Use `solve` to find the critical points. Solve the system
$$
\frac{\partial f}{\partial x} = 0, \frac{\partial f}{\partial y} = 0
$$

````python
from sympy import solve
````

````python
solns = solve([f_x, f_y], [x, y], dict=True)
````

````python
solns
````

For this course, we only need the real solutions. We have two critical points $(x, y) = (0, 0)$ and $(x, y) = (1, 1)$.

Compute the Hessian

$$
H(f) = \begin{pmatrix}
\frac{\partial^2f}{\partial x^2} & \frac{\partial^2f}{\partial y} \\
\frac{\partial^2f}{\partial y} & \frac{\partial^2f}{\partial y^2}
\end{pmatrix}
$$

````python
h = hessian(f, [x, y])
````

````python
h
````

**Result from cal 3:**

- If Hessian at the point is negative, then it is a saddle point.
- If Hessian is positive and $\frac{\partial^2f}{\partial x^2}>0$, then this point is a local min.
- If Hessian is positive and $\frac{\partial^2f}{\partial x^2}<0$, then this point is a local max.

At the point $(x, y) = (0, 0)$:

````python
# compute the determinate
from sympy import det
````

````python
# plug in x = 0, y = 0 into the hessian (function of two variables)
# function.subs({dictionary of x, y key-value})
hessian_val = det(h.subs({x: 0, y: 0}))
hessian_val
````

Since hessian = -9 < 0, the point $(0, 0)$ is a saddle point.

````python
# plugin the second point
hessian_val2 = det(h.subs({x: 1, y: 1}))
hessian_val2
````

````python
f_xx = diff(f_x, x)
f_xx
````

````python
# plugin x=1, y=1 into f_xx
a_xx = f_xx.subs({x: 1, y: 1})
print(a_xx)
````

Since hessian is positive and $f\_xx > 0$, the point $(1, 1)$ is local min.

## Solving ODEs using Sympy

**Definition:** Ordinary differential equation is just an equation that contains the derivative of a function.
Solution of an ODE is a *function* that satisfies the given equation.

**Example:** For the equation $y' -9y = 0$ we can show that the function
$$
y(t) = e^{9t}
$$
is a solution.

We have $y' = 9e^{9t}$. So
$$
y' - 9y = 9e^{9t} - 9(e^{9t}) = 0,\forall t.
$$

````python
# for solve ODEs
from sympy import symbols, Function, dsolve, Derivative, Eq
````

**Example:** Solve the ODE: $y' -9y = 0$
- Suppose $y$ is a function of $t$
- No initial conditions
- $t$ is the variable, and $y$ is a special variable (function)

````python
# define t
t = symbols("t")
````

````python
# define special variable y as function
y = symbols("y", cls=Function) # tell Sympy the y is class of function, not regular var!
````

````python
t
````

````python
y(t)
````

````python
# define the equation LHS = y' - 9y, RHS = 0
eq = Eq(Derivative(y(t), t) - 9*y(t), 0)
eq
````

````python
# solve the equation
solns = dsolve(eq, y(t))
````

````python
solns
````

**Example:** Solve the ODE:
$$
y' + 7y = 5x^3 - 2x^2 + \sin(x)
$$

````python
from sympy import symbols, Function, sin, Eq, Derivative, dsolve
````

````python
# define the x var
x = symbols("x")
````

````python
# define the y as a function!
y = symbols("y", cls=Function)
````

````python
# define the equation
# notice: y is a function, and y(x) the function of x
eq = Eq(Derivative(y(x), x) + 7 * y(x), 5 * x**3 - 2 * x**2 + sin(x))
````

````python
eq
````

````python
# solve it
solns = dsolve(eq, y(x))
````

````python
solns
````

**Example:** Solve ODEs with two initial conditions

$$
y''-2y'+5y = e^t\cos(2t), y(0) = -2, y'(0) = 3.
$$

````python
# define t
t = symbols("t")
y = symbols("y", cls=Function)
````

````python
from sympy import exp, cos
````

````python
# define the equation
eq = Eq(Derivative(y(t), t, 2) - 2*Derivative(y(t), t) + 5*y(t), exp(t) * cos(2*t))
````

````python
eq
````

````python
# solve it
y_prime = y(t).diff()
y_prime
````

````python
# solve the equation with y(0) = -2 and y'(0) = 3
solns2 = dsolve(eq, y(t), ics={y(0):-2,y_prime.subs({t: 0}): 3})
````

````python
solns2
````

## Solving a system of ODEs

**Example:** Solve the system

$$
\frac{dx}{dt} = 3x + 4y, \frac{dy}{dt} = -4x + 3y.
$$

````python
# reuse t from previous problem
# define x and y as functions
x, y = symbols("x y", cls=Function)
````

````python
# define the equations
eq1 = Eq(Derivative(x(t), t), 3*x(t) + 4*y(t))
eq2 = Eq(Derivative(y(t), t), -4*x(t) + 3*y(t))
````

````python
eq1
````

````python
eq2
````

````python
# define the system
eqs = [eq1, eq2]
````

````python
# solve the system
solns = dsolve(eqs, [x(t), y(t)])
````

````python
x_t = solns[0]
y_t = solns[1]
````

````python
x_t
````

````python
y_t
````
