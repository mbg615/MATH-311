```python
from sympy import symbols, Function, Derivative, Eq, dsolve
```

```python
# define t
t = symbols("t")
y = symbols("y", cls=Function)
```

$$
y'' - 3y' + 2y = \sin(2t) - t^3 + e^t.
$$

```python
from sympy import sin, exp
```

```python
# set equation
eq = Eq(Derivative(y(t), t, t) - 3*Derivative(y(t), t) + 2*y(t), sin(2*t) - t**3 + exp(t))
```

```python
eq
```

```python
soln = dsolve(eq, y(t))
```
