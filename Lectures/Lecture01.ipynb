**Example:** Write a function to take the input as a list of numbers and return a dictionary of the values and ranking in the list.

````python
# write a function to return the dictionary of the values and ranking
def ranking_order(x):
    """
    input:
        x: a list of number

    return:
        a dictionary of ranking
    """
    # sort the list in descending order
    y = sorted(x, reverse=True)
    # create an empty dictionary
    d = {}
    #set initial rank =1
    r = 1
    for v in y:
        if v not in d:
            d[v] = r
            r = r + 1

    return d
````

````python
# test cases
x = [2, 5, 6, 7, 4]
````

````python
ranking_order(x)
````

````python
ranking_order(x1)
````

````python
x2 = [3, 3, 4, 4, 5, 5, 5]
````

````python
ranking_order(x2)
````

**Example:** Use python to determine whether we can write each even number as the sum of two prime numbers for all even numbers less than or equal to $200.$

Write a function to check prime number.


````python
import math
````

````python
def is_prime(n):
    """
    Check if n is prime
    """
    # some special case
    if n < 2:
        return False
    if n == 2 or n == 3:
        return True
    # check the even number
    if n % 2 == 0:
        return False
    # check for odd numbers
    sqrt_n = math.sqrt(n)
    for p in range(5, int(sqrt_n) + 1, 2):
        if n % p == 0:
            # p divides n
            return False

    return True
````

````python
# generate a random int
n = random.randint(50, 300)
print(n)
print(is_prime(n))
````

Write a function to return all prime numbers from 2 to n=200

````python
def list_primes(n):
    # create an empty list
    p_list = []
    for i in range(2, n + 1):
        if is_prime(i):
            p_list.append(i)
    return p_list
````

````python
# test the function
list_primes(20)
````

# Check the conjecture up to n = 200

````python
def verify_conjecture(n=200):
    # the list of all primes up to 200
    primes = list_primes(n)
    for even_num in range(2, 201, 2):
        checked = False
        for p in primes:
            q = even_num - p
            if q in primes:
                print(f"{even_num} = {p} + {q}")
                checked = True
        if not checked:
            print(f"Counter example for even number: 2")
````

````python
verify_conjecture(20)
````
