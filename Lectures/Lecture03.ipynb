**Elementwise product:**

If $x = (x_1,x_2,\ldots,x_n)$ and $y = (y_1,y_2,\ldots,y_n)$ then
$$
x*y = (x_1y_1, x_2y_2, \ldots, x_ny_n).
$$

```python
# what is x * y?
print(x*y)
```

**Dot Product:**
If $x = (x_1,x_2,\ldots,x_n)$ and $y = (y_1,y_2,\ldots,y_n)$ then
$$
x \dot y = x_1y_1 + x_2y_2+\cdots+x_ny_n.
$$

```python
# compute the dot product
np.dot(x,y) == np.sum(x*y)
```

## Norm (len) of vectors
A norm of a vector is defined as
$$
\|x\|_{p} = (|x_1|^p+|x_2|^p+\cdots+|x_n|^p)^{1/p}
$$
In Numpy, we only need for $p=1$, $p=2$ or $p=\infty$

```python
# compute the norm of vector
v = np.array([3.0, -4.0, 0])
```

```python
l1_norm = np.linalg.norm(v, 1)
print(l1_norm)
```

```python
linf_norm = np.linalg.norm(v, np.inf)
print(linf_norm)
```

```python
# 12 norm
l2_norm = np.linalg.norm(v, 2)
print(l2_norm)
```

```python
l2_norm = np.linalg.norm(v)
print(l2_norm)
```

**Example:** Write a function to compute the cosine similarity of two vectors.
If $x = (x_1,\ldots,x_n)$ and $y = (y_1,\ldots,y_n)$ then
$$
\cos(\theta) = \frac{x\cdot y}{\|x\|\|y\|}.
$$

```python
def cos_sim(x, y):
    return (x @ y) / (np.linalg.norm(x) * np.linalg.norm(y))
```

```python
# make up vectors
cat = np.array([0.9, 0.1, 0.0])
dog = np.array([0.85, 0.2, 0.0])
car = np.array([0.0, 0.1, 0.9])
```

```python
# check the similarity between cat and dog
cos_sim(cat, dog)
```

```python
cos_sim(cat, car)
```

## Rank and determinate of matrices

```python
A = np.array([
    [1.1, 2.1, 2.5],
    [2.2, 3.1, -1.1],
    [0.5, 0.1, -0.3]
])
```

```python
rank = np.linalg.matrix_rank(A)
```

```python
print(rank)
```

```python
# compute the determinant
det = np.linalg.det(A)
print(det)
```

```python
A_inv = np.linalg.inv(A)
```

```python
print(A_inv)
```

```python
A @ A_inv
```

```python
# generate an identity matrix
I_3 = np.eye(3)
```

```python
I_3
```

```python
# Check if A@A_inv == I_3 (identity)
np.allclose(A@A_inv, I_3)
```

## Vector-valued function
Suppose we have a function $y=f(x)$ and a vector $v = (v_1,\ldots, v_n)$. We want to define $f(v)$.

$$
f(v) = (f(v_1),\ldots, f(v_1)).
$$

**Example:**
- Generate a random vector, v, (float) of size n = 200
- Use the for loop to determine the sin(v).

```python
# compute sin(v)
import math
```

```python
n = 5000
# gen random vector (either normal or uniform distribution)
v = np.random.normal(loc=0, scale=1.5, size=n)
print(v.shape)
```

```python
# use for loop
# write a function to compute sin_v
def sin(v):
    result = []
    n = len(v)
    for i in range(n):
        result.append(math.sin(v[i]))
    sin_v = np.array(result)
    return sin_v
```

```python
print(sin(v).shape)
```

**Measure the running time**

```python
import time
```

```python
# test the computing time for n from 5000 to 10000
for n in range(5_000, 10_000, 100):
    v = np.random.normal(loc=0, scale=1.5, size=n)
    star_time = time.time()
    sin_v = sin(v)
    end_time = time.time()
    print(f"The running time for n={n} is {end_time - star_time}")
```
