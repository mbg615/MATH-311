## Norm (len) of vectors
A norm of a vector is defined as
$$
\|x\|_{p} = (|x_1|^p+|x_2|^p+\cdots+|x_n|^p)^{1/p}
$$
In Numpy, we only need for $p=1$, $p=2$ or $p=\infty$

```python
# compute the norm of vector
v = np.array([3.0, -4.0, 0])
```

```python
l1_norm = np.linalg.norm(v, 1)
print(l1_norm)
```

```python
linf_norm = np.linalg.norm(v, np.inf)
print(linf_norm)
```

```python
# 12 norm
l2_norm = np.linalg.norm(v, 2)
print(l2_norm)
```

```python
l2_norm = np.linalg.norm(v)
print(l2_norm)
```

**Example:** Write a function to compute the cosine similarity of two vectors.
If $x = (x_1,\ldots,x_n)$ and $y = (y_1,\ldots,y_n)$ then
$$
\cos(\theta) = \frac{x\cdot y}{\|x\|\|y\|}.
$$

```python
def cos_sim(x, y):
    return (x @ y) / (np.linalg.norm(x) * np.linalg.norm(y))
```

```python
# make up vectors
cat = np.array([0.9, 0.1, 0.0])
dog = np.array([0.85, 0.2, 0.0])
car = np.array([0.0, 0.1, 0.9])
```

```python
# check the similarity between cat and dog
cos_sim(cat, dog)
```

```python
cos_sim(cat, car)
```

## Vector-valued function
Suppose we have a function $y=f(x)$ and a vector $v = (v_1,\ldots, v_n)$. We want to define $f(v)$.

$$
f(v) = (f(v_1),\ldots, f(v_1)).
$$

**Example:**
- Generate a random vector, v, (float) of size n = 200
- Use the for loop to determine the sin(v).

```python
# compute sin(v)
import math
```

```python
n = 5000
# gen random vector (either normal or uniform distribution)
v = np.random.normal(loc=0, scale=1.5, size=n)
print(v.shape)
```

```python
# use for loop
# write a function to compute sin_v
def sin(v):
    result = []
    n = len(v)
    for i in range(n):
        result.append(math.sin(v[i]))
    sin_v = np.array(result)
    return sin_v
```

```python
print(sin(v).shape)
```

**Measure the running time**

```python
# test the computing time for n from 5000 to 10000
for n in range(5_000, 10_000, 100):
    v = np.random.normal(loc=0, scale=1.5, size=n)
    star_time = time.time()
    sin_v = sin(v)
    end_time = time.time()
    print(f"The running time for n={n} is {end_time - star_time}")
```

Two functions in your homework: `Relu` and `softmax`.

$Relu(x) = \max(0,x)$. If $x > 0$ then $Relu(x) = x$; otherwise $Relu(x) = 0.$

```python
def relu(x):
    x = np.array(x)
    zero = np.zeros_like(x)
    return np.max(np.array([zero, x]), axis=0)
```

```python
def relu2(x):
    return max(0,x)
```

```python
# test them
x = -0.5
print(relu(x))
print(relu2(x))
```

```python
x = [1, -2]
print(relu(x))
```

```python
x = np.random.normal(loc=0, scale=2, size=10)
print(x)
```

```python
relu(x)
```

Define `softmax` function: Turn a vector into probabilies. If the value inside the vector id larger, it must have a bigger probability.

Math formula:
$$
softmax(v) = \frac{(e^{v_1},\ldots,e^{v_n})}{e^{v_1}+\cdots+e^{v_n}}
$$

```python
# define vectorized function softmax
def softmax(v):
    # if c is not a numpy array
    v = np.array(v)
    e_v = np.exp(v)
    return e_v / np.sum(e_v)
```

```python
# test the function
v = np.random.normal(loc=0, scale=2.0, size=10)
print(v)
```

```python
softmax(v).round(3)
```

**Classification problems:**

Suppose we have 3 types of fruits: apples, banana, and orange. Each fruit has a presentation as a vector of 4 dimensions. Below is the data set for `[apple, apple, banana, banana, orange, orange]`

```python
fruit_labels = ["apple", "apple", "banana", "banana", "orange", "orange"]
```

```python
fruit_vectors = np.array([
    [0.78, 0.5, 0.88, 0.65],
    [0.79, 0.39, 0.9, 0.69],
    [0.08, 0.94, 0.82, 0.78],
    [0.07, 0.96, 0.79, 0.8],
    [0.15, 0.47, 0.48, 0.91],
    [0.1, 0.41, 0.4, 0.98]
])
fruit_vectors
```

Suppose we have a new fruit (one of the 3 given) with vector

```python
new_fruit_vector = np.array([0.05, 0.93, 0.79, 0.82])
```

Determine the name of the new fruit.

```python
# define cosine similarity score
def cos_score(a,b):
    return np.dot(a,b) / (np.linalg.vector_norm(a) * np.linalg.vector_norm(b))
```

```python
a = np.random.normal(loc=0, scale=1.5, size=5)
print(a)
```

```python
b = np.random.normal(loc=0, scale=1.5, size=5)
print(b)
```

```python
# test cos_score function
score = cos_score(a,b)
print(score)
```

To identify the new fruit, we compare the cosine scores of the new vectors paring with the known fruit.

```python
first_apple = fruit_vectors[0]
print(first_apple)
```

```python
score1 = cos_score(first_apple,new_fruit_vector)
print(score1)
```

```python
# better to use loop
for i, v in enumerate(fruit_vectors):
    # i -> 0, 1, 2, 3, 4, 5
    # v -> fruit vectors in order
    score_i = cos_score(new_fruit_vector, v)
    print(f"The score of the new fruit vs {fruit_vectors[i]} is {score_i}")
```

Based on the output scores, the new fruit is banana

```python
# compute all scores at once
scores = []
for i in range(6):
    score = cos_score(new_fruit_vector, fruit_vectors[i])
    scores.append(score)
# convert list score to Numpy
scores = np.array(scores)
```

```python
scores
```

Find the index where the score is max

```python
max_idx = np.argmax(scores)
print(max_idx)
```

```python
label = fruit_labels[max_idx]
print(label)
```

## Small neural network!

$$
XW_1 + b_1 \rightarrow ReLU \rightarrow HW_2 + b_2 \rightarrow softmax \rightarrow class
$$
where $W_1, W_2$ are to matrices (weights) and two vectors $b_1, b_2$ (biases); the activation function is $ReLU$ and $softmax$ is the probability function.

```python
W_1 = np.array(
    [
        [-4.91, -0.09, 0.31],
        [6.06, -0.28, 0.17],
        [2.21, 0.45, -0.36],
        [-0.90, -0.33, 0.01],
    ]
)
print(W_1)
```

```python
W_2 = np.array(
    [
        [-7.04, 3.99, 1.16],
        [-0.36, -0.27, -0.16],
        [0.21, 0.52, -0.07],
    ]
)
print(W_2)
```

```python
b_1 = np.array([-0.55, -0.03, -0.01])
b_2 = np.array([7.13, -8.65, 1.52])
```

```python
X = new_fruit_vector
```

```python
# compute XW_1 + b_1 -> y_1
y_1 = X@W_1 + b_1
```

```python
print(y_1)
```

```python
# compute H
H = relu(y_1)
print(H)
```

```python
# compute HW_2 + b_2 -> y_2
y_2 = H@W_2 + b_2
```

```python
print(y_2)
```

```python
# apply the softmax to get the probability
prob = softmax(y_2)
print(prob.round(3))
```

Notice: The probability vector has 3 values which are the probability for the class: apple, banana and orange in this order.

probabiluity( new fruit os apple) = 0

probabiluity( new fruit os banana) = 0.998

probabiluity( new fruit os orange) = 0.002

```python
classes = ["apple", "banana", "orange"]
```

```python
max_idx = np.argmax(prob)
print(max_idx)
```

```python
fruit = classes[max_idx]
print(fruit)
```
