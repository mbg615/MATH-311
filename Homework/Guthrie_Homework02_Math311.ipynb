## 1.1 Exercise 1: Taylor Series Approximation
Use SymPy to compute the Taylor series of $f(x)$ about $0$ with the first $10$ non-zero
terms.

```python
from sympy import symbols, exp, sin, series

x = symbols("x")

f = exp(x) * sin(2 * x)

x0 = 0
n = 11

f_series = series(f, x, x0, n)
t_series = f_series.removeO()
t_series
```

Use the Taylor polynomial to approximate $f(0.4)$.

```python
f_approx = t_series.subs({x:0.4})
f_approx
```

Compute the exact value of $f(0.4)$ using `math` package.

```python
import math
f_actual = math.exp(0.4) * math.sin(2 * 0.4)
f_actual
```

Compute the absolute error between the approximation and the actual value of the function
at $x = 0.4$.

```python
err = abs(f_approx - f_actual)
err
```

The approximation is very accurate. The absolute error is only $~2.38 \times 10^{-9}$, so the Taylor series matches $e^x sin(2x)$ to 8 or 9 significant figures at $x = 0.4$. This makes sense since $x = 0.4$ is close to the expansion point, so the series converges fast and the truncated terms are very small.

## 1.2 Exercise 2: Ordinary Differential Equations

### 1.2.1 Part A: Second-Order ODE with Forcing Term
Solve the non-homogeneous equation:
$$
y'' - 3y' + 2y = \sin(2t) - t^3 + e^t
$$

```python
from sympy import symbols, Function, Eq, Derivative, dsolve

t = symbols("t")
y = symbols("y", cls=Function)

eq = Eq(Derivative(y(t), t, 2) - 3 * Derivative(y(t), t, 1) + 2 * y(t), sin(2 * t) - t ** 3 + exp(t))
solns = dsolve(eq, y(t))
solns
```

### 1.2.2 Part B: Second-Order ODE with Initial Conditions
Solve the equation in Part A subject to:
$$
y(0) = 0, \quad y'(0) = 1
$$

```python
solns = dsolve(eq, y(t), ics={y(t).subs(t, 0): 0, y(t).diff(t).subs(t, 0): 1})
solns
```

### 1.2.3 Part C: System of First-Order ODEs (Complex Eigenvalues)
Solve the system:
$$
\begin{cases}
x'(t) = x - 2y + sin(2t) \\
y'(t) = 2x + y + t^2
\end{cases}
$$

```python
t = symbols("t")
x = symbols("x", cls=Function)
y = symbols("y", cls=Function)

eq1 = Eq(x(t).diff(t), x(t) - 2 * y(t) + sin(2 * t))
eq2 = Eq(y(t).diff(t), 2 * x(t) + y(t) + t**2)
eqs = [eq1, eq2]

solns = dsolve(eqs, [x(t), y(t)])
x_t = solns[0]
y_t = solns[1]

print("General Solution")
display(x_t, y_t)
```

### 1.2.4 Part D: System of ODEs with Initial Conditions
Solve the system in Part C subject to:
$$
x(0) = 1, \quad y(0) = 0
$$
Clearly distinguish between general and IVP solutions.

```python
solns = dsolve(eqs, [x(t), y(t)], ics={x(0): 1, y(0): 0})
x_t = solns[0]
y_t = solns[1]

print("IVP Solution:")
display(x_t, y_t)
```

## 1.3 Exercise 3: Critical Points and Hessian Classification
Consider the function
$$
f(x, y) = x^4 + y^4 - 4xy + 1
$$
Compute the first-order partial derivatives $f_x$ and $f_y$.

```python
from sympy import symbols, diff

x, y = symbols("x y")
f = x**4 + y**4 - 4*x*y + 1

f_x = diff(f, x)
f_y = diff(f, y)

display(f_x, f_y)
```

Solve $\nabla f = 0$ to find all critical points.

```python
from sympy import solve
critical_points = solve([f_x, f_y], [x, y])
critical_points
```

Compute the Hessian matrix of $f(x, y)$.

```python
from sympy import hessian

h = hessian(f, [x, y])
h
```

Evaluate the Hessian at each critical point.

```python
from sympy import det

display(det(h.subs({x: -1, y: -1})),
        det(h.subs({x: 0, y: 0})),
        det(h.subs({x: 1, y: 1})))
```

Use the second derivative test to classify each point as:
- Local maximum
- Local minimum
- Saddle point

```python
f_xx = diff(f_x, x)
display(f_xx.subs({x: -1, y: -1}),
        f_xx.subs({x: 0, y: 0}),
        f_xx.subs({x: 1, y: 1}))
```

$(-1,-1)$ Local minimum \
$(0,0)$ Saddle Point \
$(1,1)$ Local minimum
