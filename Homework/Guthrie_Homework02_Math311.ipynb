## 1.1 Exercise 1: Taylor Series Approximation
Use SymPy to compute the Taylor series of $f(x)$ about $0$ with the first $10$ non-zero
terms.

```python
from sympy import symbols, exp, sin, series

x = symbols("x")

f = exp(x) * sin(2 * x)

x0 = 0
n = 11

f_series = series(f, x, x0, n)
t_series = f_series.removeO()
t_series
```

Use the Taylor polynomial to approximate $f(0.4)$.

```python
f_approx = t_series.subs({x:0.4})
f_approx
```

Compute the exact value of $f(0.4)$ using `math` package.

```python
import math
f_actual = math.exp(0.4) * math.sin(2 * 0.4)
f_actual
```

Compute the absolute error between the approximation and the actual value of the function
at $x = 0.4$.

```python
err = abs(f_approx - f_actual)
err
```

The approximation is very accurate. The absolute error is only $~2.38 \times 10^{-9}$, so the Taylor series matches $e^x sin(2x)$ to 8 or 9 significant figures at $x = 0.4$. This makes sense since $x = 0.4$ is close to the expansion point, so the series converges fast and the truncated terms are tiny.

## 1.2 Exercise 2: Ordinary Differential Equations

### 1.2.1 Part A: Second-Order ODE with Forcing Term
Solve the non-homogeneous equation:
$$
y'' - 3y' + 2y = \sin(2t) - t^3 + e^t
$$

```python
from sympy import symbols, Function, Eq, Derivative, dsolve

t = symbols("t")
y = symbols("y", cls=Function)

eq = Eq(Derivative(y(t), t, 2) - 3 * Derivative(y(t), t, 1) + 2 * y(t), sin(2 * t) - t ** 3 + exp(t))
solns = dsolve(eq, y(t))
solns
```

### 1.2.2 Part B: Second-Order ODE with Initial Conditions
Solve the equation in Part A subject to:
$$
y(0) = 0, \quad y'(0) = 1
$$

```python
solns = dsolve(eq, y(t), ics={y(t).subs(t, 0): 0, y(t).diff(t).subs(t, 0): 1})
solns
```
