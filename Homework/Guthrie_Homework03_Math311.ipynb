## 1.1 Exercise 1: Column Replacement and Determinants (Cramer’s Rule Style)

### 1.1.1 Replace the i-th column of a matrix
Given a square matrix $A$ of size $n \times n$, a vector $v$ of size $n$ and an index $i$, where $0 \leq i \leq n - 1$.
Write a function that replaces the $i$-th column of $A$ with $v$ and returns a new matrix.

```python
import numpy as np
from sympy.integrals.risch import frac_in


def replace_column(A, v, i):
    cpy = A.copy().astype(float)
    cpy[:, i] = v
    return cpy
```

### 1.1.3 Cramer’s Rule for Solving a System of Equations
Write a Python function that uses Cramer’s Rule to solve the system $Ax = b$. The function
should take $A$ and $b$ as inputs and return the solution vector $x$.

```python
A = np.array([[2, 1, 1],
              [1, 3, 2],
              [1, 0, 0]])

b = np.array([4, 5, 6])
```

```python
def cramer_solver(A, b):
    A_det = np.linalg.det(A)
    x = []
    for i in range(A.shape[0]):
        A_i = replace_column(A, b, i)
        A_i_det = np.linalg.det(A_i)
        x.append(A_i_det / A_det)
    return x

x1 = cramer_solver(A, b)
print(x1)
```

Use the NumPy solver np.linalg.solve() to solve the same system.

```python
x2 = np.linalg.solve(A, b)
print(x2)
```

Compare the two solutions to verify that your implementation of Cramer’s Rule is correct.

```python
print(np.isclose(x1, x2))
```

**Write a Python function that computes the cosine similarity between two vectors**

```python
def cosine_similarity(u,v):
    u_v_dot = np.dot(u,v)
    u_norm = np.linalg.norm(u)
    v_norm = np.linalg.norm(v)
    cos_sim = u_v_dot / (u_norm * v_norm)
    return cos_sim
```

### 1.2.3 A Basic Neural Network

```python
fruit_vectors = np.array([
    [0.78, 0.5, 0.88, 0.65],
    [0.79, 0.39, 0.9, 0.69],
    [0.08, 0.94, 0.82, 0.78],
    [0.07, 0.96, 0.79, 0.8],
    [0.15, 0.47, 0.48, 0.91],
    [0.1, 0.41, 0.4, 0.98]
])

W1 = np.array(
    [
        [-4.91, -0.09, 0.31],
        [6.06, -0.28, 0.17],
        [2.21, 0.45, -0.36],
        [-0.90, -0.33, 0.01],
    ]
)

W2 = np.array(
    [
        [-7.04, 3.99, 1.16],
        [-0.36, -0.27, -0.16],
        [0.21, 0.52, -0.07],
    ]
)

b1 = np.array([-0.55, -0.03, -0.01])
b2 = np.array([7.13, -8.65, 1.52])

classes = ["apple", "banana", "orange"]
```

Write a function that takes a fruit vector 𝑋 as input and performs all the computations:
$XW1 + b1 \to ReLU \to HW2 + b2 \to softmax$

```python
def relu(x):
    return np.maximum(0, x)

def softmax(x):
    return np.exp(x) / np.sum(np.exp(x))

def fruit_nn_classifier(X, W1, W2, b1, b2, classes):
    y1 = np.dot(X, W1) + b1
    H = relu(y1)
    y2 = np.dot(H, W2) + b2
    prob = softmax(y2)
    max_idx = np.argmax(prob)
    return classes[max_idx]
```

```python
for fruit in fruit_vectors:
    pred_name = fruit_classifier(fruit)
    print(f"Predicted: {pred_name} for {fruit}")
```
