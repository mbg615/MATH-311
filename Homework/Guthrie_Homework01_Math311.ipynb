# 1.1 Exercise 1: Fibonacci Sequence

**Function to return the list of first 𝑛 Fibonacci numbers:**

```python
def fibonacci(n):
    result = []
    a0 = a1 = 1
    result.extend([a0, a1])
    while n - 2 > 0:
        a0, a1 = a1, a0 + a1
        result.append(a1)
        n -= 1

    return result
```

```python
for n in range(10, 50, 5):
    result = fibonacci(n)
    if len(result) == n:
        print("passed")
```

**Function to return the 𝑛-th number in the Fibonacci sequence.**

```python
def nth_fibonacci(n):
    a0 = a1 = 1
    while n - 1 > 0:
        a0, a1 = a1, a0 + a1
        n = n - 1

    return a1
```

```python
for n in range(30, 50, 5):
    nth_number = nth_fibonacci(n)
    print(nth_number)
```

```python
for n in range(0, 10, 1):
    print(nth_fibonacci(n))
```

# 1.2 Exercise 2: Ranking Test Scores Using a Dictionary

```python
def rank_scores(scores:list) -> list:
    sorted_scores = sorted(scores, reverse=True)
    scores_dict = {}
    for rank, score in enumerate(sorted_scores):
        if score not in scores_dict:
            scores_dict[score] = rank + 1

    return [scores_dict[score] for score in scores]
```

```python
test_list = [32, 87, 75, 75, 50, 32]
result = rank_scores(test_list)
print(result)
```

# 1.3 Exercise 3: Normalize a List of Numbers

```python
def normalize_nums(nums:list) -> list:
    if not nums:
        return []
    x_max = max(nums)
    x_min = min(nums)
    if x_max == x_min:
        return [0.0 for x in nums]
    return [(x - x_min) / (x_max - x_min) for x in nums]
```

```python
test_1 = [2, 4, 10, 6, 8, 4]
print(f"{test_1} normalized: {normalize_nums(test_1)}")
test_2 = [5, 5, 5, 5]
print(f"{test_2} normalized: {normalize_nums(test_2)}")
test_3 = []
print(f"{test_3} normalized: {normalize_nums(test_3)}")
```

# 1.4 Exercise 4: Understand, Debug, and Test a Python Function

```python
def is_harshad(n):
    """Determine whether or not a number is a Harshad number."""
    n_digits = list(str(n))
    sum_digits = 0
    for x in n_digits:
        sum_digits += x
    return n % sum_digits == 0
```

## 1.4.1 Understand the Function

`def is_harshad(n):` Define a function named "is_harshad" with $n$ as an input. \
`n_digits = list(str(n))` Interpret $n$ as a string and create a list of its characters. \
`sum_digits = 0` Create the variable "sum_digits" and set it to $0$. \
`for x in n_digits:` Loop through the characters in "n_digits" one at a time assigning the current character to $x$. \
`sum_digits += x` Add $x$ to "sum_digits". \
`return n % sum_digits == 0` Return "True" if "n" is divisible by "sum_digits" otherwise return "False".

### Tracing `is_harshad(21)`
When `is_harshad(n)` is called with the value $21$, the argument `n` is set to the integer value $21$.
The first line, `n_digits = list(str(n))`, converts `n` to a string using `str(n)`. This turns the integer `21` into the string "21". The `list()` function then breaks that string into a list of characters, producing ['2', '1']. So, `n_digits` is a list of strings, where each element is a single-character string.

Next `sum_digits` is created and initialized to $0$.

The `for` loop then iterates over the values of `n_digits`, assigning them to `x`. In the first iteration `x` $ = $ '2', and in the second `x` $ = $ '1'.

Inside the loop `sum_digits += x` tries to add `x` to `sum_digits`. However, because `x` is a string this raises a `TypeError` because python cannot add a string, and an integer. The function stops here and won't reach the `return` statement.

Assuming `x` is an integer, the loop would have correctly summed the digits. After the first iteration `sum_digits` $ = 2$ and after the second, `sum_digits` $ = 3$.

Finally, the result of `n % sum_digits` would be calculated. Using `%` to find the remainder when `n` is divided by `sum_digits`. Because $21 / 3 = 7$ and the remainder is zero, the function returns `True`.

## 1.4.2 Debug the Function

This code contains a `TypeError` on line 6. This error is raised because the code attempts to add `x` which is a string, to `sum_digits` which is an integer. This is because `x` is assigned to the values in the list `n_digits` which are strings due to `list(str(n))`. To fix this we can simply convert `x` to an integer before adding it to `sum_digits`. The final code can be seen below. This fixes the issue since both `sum_digits` and `x` are now integers, so Python can add them together.

```python
def is_harshad(n):
    """Determine whether or not a number is a Harshad number."""
    n_digits = list(str(n))
    sum_digits = 0
    for x in n_digits:
        sum_digits += int(x)
    return n % sum_digits == 0
```

## 1.4.3 Test the Function

```python
test_nums = [1, 10, 12, 15, 18, 20, 21, 22, 24, 100, 101, 111]
for n in test_nums:
    if is_harshad(n):
        print(f"{n:<3} is a Harshad number")
    else:
        print(f"{n:<3} is not a Harshad number.")
```
